Citi Bike demand varies across the day and across stations. We use September 2026 trips with both endpoints in Manhattan to examine when rides leave and where recorded arrivals exceed or fall short of departures. The comparison uses 21 ordinary weekdays and 8 weekend days. Labor Day is excluded. Counts are divided by the number of days in each group so the two day types can be compared on a daily basis.

The [data documentation](../manhattan-data/README.md) describes the official Citi Bike trip archive, the NYC Department of City Planning boundary, and the processing steps.

```python
from charts import basemap, embed, hourly_departures, station_net_flow, DAYS
```

## Q1 How does the timing of departures differ by day type?

The horizontal axis shows the local hour when a trip started. Each line shows the average number of departures in that hour on an ordinary weekday or weekend day. Move across the chart or use the hour slider to compare daily averages, monthly counts, and the number of days behind each average.

```python
embed("q1", hourly_departures(), "Q1 hourly Citi Bike departures", height=510)
```

### What the hourly comparison shows

Ordinary weekdays have two distinct peaks, at 08:00 and 17:00, with about **8,545** and **11,409** departures per day in those respective hours. Weekend trips build up more gradually and reach their highest hourly average at 15:00, at about **5,440** departures per day. At noon the two day types are close, with **4,783** weekday and **4,817** weekend departures per day. Summed over all 24 hours, the average ordinary weekday has about **105,359** departures, compared with **67,990** on a weekend day.

The higher weekday total is concentrated around the morning and late-afternoon peaks. The shape is consistent with commuting schedules, although the trip records contain no trip-purpose field.

## Q3 Where does station net flow differ by day type?

Each mark represents a station on two maps with the same extent and size scale. Blue circles represent positive net inflow and orange triangles represent negative net inflow. Symbol area encodes the absolute difference between daily arrivals and departures. The small gray dot indicates a zero difference. Click a map symbol or choose a station in the menu to highlight the same location on both maps and compare its values.

We use the start date for each departure and the end date for each arrival. Events must occur in September and outside Labor Day. The data covers trips starting in September, so August departures arriving in September fall outside the sample.

Before plotting, we combine **34 pairs** of station IDs that differ only by a trailing zero and have the same station name and coordinates within two metres. This produces **678 station entries** from 712 raw IDs. For example, the two IDs for E 17 St & Broadway appear to have opposite weekday imbalances until their events are added together. The source tables retain their original IDs and the chart performs the documented reconciliation.

```python
stations, aliases = station_net_flow()
embed("q3", {"stations": stations, "geography": basemap(), "days": DAYS},
      "Q3 station net flow on weekdays and weekends", height=685)
```

### What the station maps show

**Norfolk St & Broome St** has the largest average net inflow in both groups, about **+28.05** trips per ordinary weekday and **+28.50** per weekend day. **E 39 St & Lexington Ave** has the largest net outflow on ordinary weekdays (**−28.62** per day), compared with **−1.25** on weekends. The largest weekend net outflow is at **FDR Drive & E 35 St** (**−23.25** per day), where the weekday balance is **+4.67** per day. The two maps show a persistent inflow at the Norfolk Street station alongside a shift in the location of the strongest recorded outflow.

The measures describe trip arrivals and departures. They do not include operator rebalancing, dock capacity, or station inventory, so they cannot establish when a station ran out of bikes or docks.
